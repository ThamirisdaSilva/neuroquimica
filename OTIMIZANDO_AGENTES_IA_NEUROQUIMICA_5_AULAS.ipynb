# OTIMIZANDO AGENTES DE IA — CASO NEUROQUÍMICA

Este notebook será utilizado durante as cinco aulas da disciplina. Cada etapa produz um arquivo que será retomado na aula seguinte.

O NeuroQuímica é um aplicativo educacional simples: o estudante faz uma pergunta sobre neurociência ou química e recebe uma explicação. O Colab será usado apenas para organizar os testes, registrar os resultados e apoiar as decisões de melhoria.

### Como utilizar

1. Execute primeiro a célula de preparação abaixo.
2. Em cada aula, execute somente o bloco correspondente.
3. Preencha os formulários e use os botões para salvar os registros.
4. Baixe o arquivo produzido ao final da aula.
5. Guarde os arquivos, pois eles serão utilizados nas etapas seguintes.

Nenhuma etapa utiliza API paga ou exige alteração no código.


```python
# PREPARAÇÃO DO NOTEBOOK — execute esta célula no início de cada aula
import io
import os
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from google.colab import files

def ler_csv_colab(conteudo):
    """Lê os CSVs gerados pelo notebook, preservando os textos e acentos."""
    return pd.read_csv(io.BytesIO(conteudo), sep=';', encoding='utf-8-sig', dtype=str).fillna('')

def baixar_csv(df, nome):
    df.to_csv(nome, sep=';', index=False, encoding='utf-8-sig')
    files.download(nome)

def valor_texto(valor):
    return '' if pd.isna(valor) else str(valor)

print('✅ Notebook preparado. Vá até a aula que será realizada hoje.')
```

## AULA 1 — PLANEJANDO OS TESTES DO NEUROQUÍMICA

Nesta aula, você vai criar perguntas para avaliar o NeuroQuímica. Cada caso deverá representar uma situação de estudo e registrar o que esperamos encontrar em uma resposta adequada.

**Entregável:** `PLANO_TESTES_NEUROQUIMICA.csv`

### Tipos de caso

- **Dúvida conceitual:** pergunta direta sobre um conceito.
- **Comparação:** pergunta sobre diferenças entre conceitos.
- **Aplicação:** situação que exige relacionar conhecimentos.
- **Pergunta ambígua:** pergunta incompleta ou pouco clara.
- **Fora do escopo:** pedido que o aplicativo educacional não deve atender.

### Exemplo de orientação

**Situação:** aluno revisando transmissão sináptica  
**Pergunta:** O que é uma sinapse química?  
**Comportamento esperado:** explicar a comunicação entre células e o papel dos neurotransmissores com linguagem acessível.  
**Critérios:** correção, clareza e atendimento ao pedido.  
**Prioridade:** alta.

O exemplo acima não será incluído automaticamente no arquivo.


```python
# AULA 1 — formulário para criação do plano de testes
if 'a1_casos' not in globals():
    a1_casos = []

a1_tipo = widgets.Dropdown(
    options=['Dúvida conceitual', 'Comparação', 'Aplicação', 'Pergunta ambígua', 'Fora do escopo'],
    description='Tipo:', style={'description_width': '150px'}, layout=widgets.Layout(width='700px'))
a1_situacao = widgets.Text(
    description='Situação:', placeholder='Ex.: aluno revisando sinapses',
    style={'description_width': '150px'}, layout=widgets.Layout(width='900px'))
a1_pergunta = widgets.Textarea(
    description='Pergunta:', placeholder='Pergunta que será feita ao NeuroQuímica',
    style={'description_width': '150px'}, layout=widgets.Layout(width='900px', height='80px'))
a1_esperado = widgets.Textarea(
    description='Esperado:', placeholder='O que uma resposta adequada precisa apresentar?',
    style={'description_width': '150px'}, layout=widgets.Layout(width='900px', height='100px'))
a1_criterios = widgets.SelectMultiple(
    options=['Correção', 'Clareza', 'Atendimento ao pedido', 'Adequação ao público', 'Reconhecimento de limites'],
    description='Critérios:', style={'description_width': '150px'}, layout=widgets.Layout(width='700px', height='120px'))
a1_prioridade = widgets.Dropdown(
    options=['Alta', 'Média', 'Baixa'], value='Média', description='Prioridade:',
    style={'description_width': '150px'}, layout=widgets.Layout(width='500px'))
a1_justificativa = widgets.Textarea(
    description='Justificativa:', placeholder='Por que este caso tem essa prioridade?',
    style={'description_width': '150px'}, layout=widgets.Layout(width='900px', height='70px'))
a1_confirmar_limpeza = widgets.Checkbox(value=False, description='Confirmo que desejo apagar todos os casos')

a1_saida_msg = widgets.Output()
a1_saida_tabela = widgets.Output()
a1_saida_revisao = widgets.Output()

def a1_atualizar():
    with a1_saida_tabela:
        clear_output()
        if a1_casos:
            display(pd.DataFrame(a1_casos))
        else:
            print('Nenhum caso foi adicionado.')
    with a1_saida_revisao:
        clear_output()
        if not a1_casos:
            return
        df = pd.DataFrame(a1_casos)
        print('REVISÃO DO PLANO')
        print(f'Total de casos: {len(df)}')
        print('\nCasos por tipo:')
        print(df['tipo_situacao'].value_counts().to_string())
        print('\nCasos por prioridade:')
        print(df['prioridade'].value_counts().to_string())
        if 'Alta' not in df['prioridade'].values:
            print('\n⚠️ Inclua ao menos um caso de prioridade alta.')
        if len(df['tipo_situacao'].unique()) == 1 and len(df) > 1:
            print('\n⚠️ Todos os casos são do mesmo tipo. Procure variar as situações.')
        if len(df) < 5:
            print(f'\n💡 Procure criar pelo menos cinco casos. Faltam {5-len(df)}.')

def a1_limpar_campos():
    a1_situacao.value = ''
    a1_pergunta.value = ''
    a1_esperado.value = ''
    a1_criterios.value = ()
    a1_justificativa.value = ''

def a1_adicionar(_):
    with a1_saida_msg:
        clear_output()
        if not a1_situacao.value.strip():
            print('❌ Preencha a situação de uso.')
            return
        if not a1_pergunta.value.strip():
            print('❌ Preencha a pergunta.')
            return
        if not a1_esperado.value.strip():
            print('❌ Preencha o comportamento esperado.')
            return
        if not a1_criterios.value:
            print('❌ Selecione pelo menos um critério.')
            return
        identificador = f'T{len(a1_casos)+1:02d}'
        a1_casos.append({
            'id_caso': identificador,
            'tipo_situacao': a1_tipo.value,
            'situacao': a1_situacao.value.strip(),
            'pergunta': a1_pergunta.value.strip(),
            'comportamento_esperado': a1_esperado.value.strip(),
            'criterios': ', '.join(a1_criterios.value),
            'prioridade': a1_prioridade.value,
            'justificativa': a1_justificativa.value.strip()
        })
        print(f'✅ Caso {identificador} adicionado.')
        a1_limpar_campos()
        a1_atualizar()

def a1_remover(_):
    with a1_saida_msg:
        clear_output()
        if not a1_casos:
            print('Não há casos para remover.')
            return
        removido = a1_casos.pop()
        print(f'🗑️ Caso {removido["id_caso"]} removido.')
        a1_atualizar()

def a1_limpar_todos(_):
    with a1_saida_msg:
        clear_output()
        if not a1_confirmar_limpeza.value:
            print('⚠️ Marque a confirmação antes de apagar todos os casos.')
            return
        a1_casos.clear()
        a1_confirmar_limpeza.value = False
        print('🗑️ Todos os casos foram apagados.')
        a1_atualizar()

def a1_baixar(_):
    with a1_saida_msg:
        clear_output()
        if not a1_casos:
            print('❌ Adicione pelo menos um caso antes de baixar.')
            return
        baixar_csv(pd.DataFrame(a1_casos), 'PLANO_TESTES_NEUROQUIMICA.csv')
        print('✅ Plano de testes criado.')

a1_btn_adicionar = widgets.Button(description='ADICIONAR CASO', button_style='success')
a1_btn_remover = widgets.Button(description='REMOVER ÚLTIMO', button_style='warning')
a1_btn_limpar = widgets.Button(description='LIMPAR CASOS', button_style='danger')
a1_btn_baixar = widgets.Button(description='BAIXAR PLANO', button_style='info')
a1_btn_adicionar.on_click(a1_adicionar)
a1_btn_remover.on_click(a1_remover)
a1_btn_limpar.on_click(a1_limpar_todos)
a1_btn_baixar.on_click(a1_baixar)

display(widgets.VBox([
    a1_tipo, a1_situacao, a1_pergunta, a1_esperado, a1_criterios,
    a1_prioridade, a1_justificativa,
    widgets.HBox([a1_btn_adicionar, a1_btn_remover]),
    a1_confirmar_limpeza, a1_btn_limpar, a1_saida_msg,
    widgets.HTML('<h4>Casos registrados</h4>'), a1_saida_tabela,
    widgets.HTML('<h4>Revisão</h4>'), a1_saida_revisao, a1_btn_baixar
]))
a1_atualizar()
```

## AULA 2 — EXECUTANDO OS TESTES E REGISTRANDO FALHAS

Nesta aula, você vai fazer as perguntas planejadas ao NeuroQuímica. Copie cada resposta para o formulário e avalie o resultado com os critérios definidos na Aula 1.

**Arquivo de entrada:** `PLANO_TESTES_NEUROQUIMICA.csv`  
**Entregável:** `BASELINE_NEUROQUIMICA.csv`

O baseline registra o comportamento da primeira versão avaliada. Ele será nossa referência nas próximas aulas.


```python
# AULA 2 — importar o plano e registrar os resultados
a2_plano = pd.DataFrame()
a2_baseline = pd.DataFrame()

a2_saida_upload = widgets.Output()
a2_saida_form = widgets.Output()
a2_saida_msg = widgets.Output()
a2_saida_resumo = widgets.Output()

a2_colunas_plano = [
    'id_caso', 'tipo_situacao', 'situacao', 'pergunta',
    'comportamento_esperado', 'criterios', 'prioridade', 'justificativa'
]
a2_colunas_resultado = [
    'versao_avaliada', 'resposta_observada', 'avaliacao',
    'evidencia', 'descricao_falha', 'observacoes'
]

def a2_resumo():
    with a2_saida_resumo:
        clear_output()
        if a2_baseline.empty:
            return
        avaliados = (a2_baseline['avaliacao'].str.strip() != '').sum()
        pendentes = len(a2_baseline) - avaliados
        print('RESUMO DA EXECUÇÃO')
        print(f'Casos avaliados: {avaliados}')
        print(f'Casos pendentes: {pendentes}')
        preenchidos = a2_baseline[a2_baseline['avaliacao'].str.strip() != '']
        if not preenchidos.empty:
            print('\nResultados:')
            print(preenchidos['avaliacao'].value_counts().to_string())
        criticos = a2_baseline[
            (a2_baseline['prioridade'] == 'Alta') &
            (a2_baseline['avaliacao'] == 'Não atende')
        ]
        if not criticos.empty:
            print('\n⚠️ Casos de prioridade alta que não atendem: ' + ', '.join(criticos['id_caso']))

def a2_criar_formulario():
    with a2_saida_form:
        clear_output()
        if a2_baseline.empty:
            return
        a2_id = widgets.Dropdown(options=a2_baseline['id_caso'].tolist(), description='Caso:')
        a2_contexto = widgets.HTML()
        a2_versao = widgets.Text(value='v1', description='Versão:', style={'description_width': '150px'})
        a2_resposta = widgets.Textarea(description='Resposta:', style={'description_width': '150px'}, layout=widgets.Layout(width='950px', height='160px'))
        a2_avaliacao = widgets.Dropdown(options=['', 'Atende', 'Atende parcialmente', 'Não atende'], description='Avaliação:', style={'description_width': '150px'})
        a2_evidencia = widgets.Textarea(description='Evidência:', placeholder='Trecho ou aspecto que sustenta sua avaliação', style={'description_width': '150px'}, layout=widgets.Layout(width='950px', height='90px'))
        a2_falha = widgets.Textarea(description='Falha:', placeholder='Descreva o problema sem tentar explicar sua causa', style={'description_width': '150px'}, layout=widgets.Layout(width='950px', height='90px'))
        a2_observacoes = widgets.Textarea(description='Observações:', style={'description_width': '150px'}, layout=widgets.Layout(width='950px', height='70px'))

        def carregar_caso(change=None):
            linha = a2_baseline[a2_baseline['id_caso'] == a2_id.value].iloc[0]
            a2_contexto.value = (
                f'<div style="padding:10px;border:1px solid #ccc">'
                f'<b>Situação:</b> {valor_texto(linha["situacao"])}<br>'
                f'<b>Pergunta:</b> {valor_texto(linha["pergunta"])}<br>'
                f'<b>Comportamento esperado:</b> {valor_texto(linha["comportamento_esperado"])}<br>'
                f'<b>Critérios:</b> {valor_texto(linha["criterios"])}<br>'
                f'<b>Prioridade:</b> {valor_texto(linha["prioridade"])}</div>'
            )
            a2_versao.value = valor_texto(linha['versao_avaliada']) or 'v1'
            a2_resposta.value = valor_texto(linha['resposta_observada'])
            a2_avaliacao.value = valor_texto(linha['avaliacao'])
            a2_evidencia.value = valor_texto(linha['evidencia'])
            a2_falha.value = valor_texto(linha['descricao_falha'])
            a2_observacoes.value = valor_texto(linha['observacoes'])

        def salvar(_):
            with a2_saida_msg:
                clear_output()
                if not a2_resposta.value.strip():
                    print('❌ Cole a resposta do NeuroQuímica.')
                    return
                if not a2_avaliacao.value:
                    print('❌ Selecione uma avaliação.')
                    return
                if not a2_evidencia.value.strip():
                    print('❌ Registre uma evidência para sua avaliação.')
                    return
                if a2_avaliacao.value != 'Atende' and not a2_falha.value.strip():
                    print('❌ Descreva a falha encontrada.')
                    return
                indice = a2_baseline.index[a2_baseline['id_caso'] == a2_id.value][0]
                a2_baseline.loc[indice, a2_colunas_resultado] = [
                    a2_versao.value.strip() or 'v1', a2_resposta.value.strip(),
                    a2_avaliacao.value, a2_evidencia.value.strip(),
                    a2_falha.value.strip(), a2_observacoes.value.strip()
                ]
                print(f'✅ Avaliação do caso {a2_id.value} salva.')
                a2_resumo()

        def baixar(_):
            with a2_saida_msg:
                clear_output()
                pendentes = (a2_baseline['avaliacao'].str.strip() == '').sum()
                if pendentes:
                    print(f'⚠️ O arquivo será baixado com {pendentes} caso(s) pendente(s).')
                baixar_csv(a2_baseline, 'BASELINE_NEUROQUIMICA.csv')
                print('✅ Baseline criado.')

        a2_id.observe(carregar_caso, names='value')
        a2_btn_salvar = widgets.Button(description='SALVAR AVALIAÇÃO', button_style='success')
        a2_btn_baixar = widgets.Button(description='BAIXAR BASELINE', button_style='info')
        a2_btn_salvar.on_click(salvar)
        a2_btn_baixar.on_click(baixar)
        carregar_caso()
        display(widgets.VBox([
            a2_id, a2_contexto, a2_versao, a2_resposta, a2_avaliacao,
            a2_evidencia, a2_falha, a2_observacoes,
            widgets.HBox([a2_btn_salvar, a2_btn_baixar]), a2_saida_msg, a2_saida_resumo
        ]))
        a2_resumo()

def a2_importar(_):
    global a2_plano, a2_baseline
    with a2_saida_upload:
        clear_output()
        enviados = files.upload()
        if not enviados:
            print('Nenhum arquivo selecionado.')
            return
        nome, conteudo = next(iter(enviados.items()))
        try:
            df = ler_csv_colab(conteudo)
        except Exception as erro:
            print(f'❌ Não foi possível abrir o arquivo: {erro}')
            return
        faltantes = [c for c in a2_colunas_plano if c not in df.columns]
        if faltantes:
            print('❌ O arquivo não parece ser o plano da Aula 1.')
            print('Colunas ausentes: ' + ', '.join(faltantes))
            return
        a2_plano = df[a2_colunas_plano].copy()
        a2_baseline = a2_plano.copy()
        for coluna in a2_colunas_resultado:
            a2_baseline[coluna] = 'v1' if coluna == 'versao_avaliada' else ''
        print(f'✅ {len(a2_baseline)} casos carregados de {nome}.')
        display(a2_plano)
        a2_criar_formulario()

a2_btn_importar = widgets.Button(description='CARREGAR PLANO DE TESTES', button_style='primary')
a2_btn_importar.on_click(a2_importar)
display(widgets.VBox([a2_btn_importar, a2_saida_upload, a2_saida_form]))
```

## AULA 3 — DIAGNOSTICANDO AS CAUSAS DAS FALHAS

Nesta aula, você vai selecionar uma falha registrada no baseline. Primeiro será descrito o que aconteceu. Depois serão levantadas possíveis explicações e definida uma pequena verificação.

**Arquivo de entrada:** `BASELINE_NEUROQUIMICA.csv`  
**Entregável:** `DIAGNOSTICO_NEUROQUIMICA.csv`

**Sintoma** é o problema observado. **Hipótese** é uma possível explicação. **Verificação** é uma pequena ação usada para descobrir se a hipótese faz sentido.


```python
# AULA 3 — selecionar uma falha e registrar o diagnóstico
if 'a3_diagnosticos' not in globals():
    a3_diagnosticos = []
a3_falhas = pd.DataFrame()

a3_saida_upload = widgets.Output()
a3_saida_form = widgets.Output()
a3_saida_msg = widgets.Output()
a3_saida_tabela = widgets.Output()

def a3_atualizar_tabela():
    with a3_saida_tabela:
        clear_output()
        if a3_diagnosticos:
            display(pd.DataFrame(a3_diagnosticos))
        else:
            print('Nenhum diagnóstico foi registrado.')

def a3_criar_formulario():
    with a3_saida_form:
        clear_output()
        a3_id_caso = widgets.Dropdown(options=a3_falhas['id_caso'].tolist(), description='Caso:')
        a3_contexto = widgets.HTML()
        a3_sintoma = widgets.Textarea(description='Sintoma:', style={'description_width': '170px'}, layout=widgets.Layout(width='950px', height='80px'))
        a3_evidencia = widgets.Textarea(description='Evidência:', style={'description_width': '170px'}, layout=widgets.Layout(width='950px', height='80px'))
        a3_h1 = widgets.Textarea(description='Hipótese 1:', style={'description_width': '170px'}, layout=widgets.Layout(width='950px', height='70px'))
        a3_h2 = widgets.Textarea(description='Hipótese 2:', style={'description_width': '170px'}, layout=widgets.Layout(width='950px', height='70px'))
        a3_h3 = widgets.Textarea(description='Hipótese 3:', placeholder='Opcional', style={'description_width': '170px'}, layout=widgets.Layout(width='950px', height='70px'))
        a3_escolha = widgets.Dropdown(options=['Hipótese 1', 'Hipótese 2', 'Hipótese 3'], description='Escolhida:', style={'description_width': '170px'})
        a3_justificativa = widgets.Textarea(description='Por que escolher?', style={'description_width': '170px'}, layout=widgets.Layout(width='950px', height='70px'))
        a3_verificacao = widgets.Textarea(description='Verificação:', style={'description_width': '170px'}, layout=widgets.Layout(width='950px', height='80px'))
        a3_resultado = widgets.Textarea(description='Resultado esperado:', style={'description_width': '170px'}, layout=widgets.Layout(width='950px', height='80px'))
        a3_prioridade = widgets.Dropdown(options=['Alta', 'Média', 'Baixa'], value='Alta', description='Prioridade:', style={'description_width': '170px'})

        def exibir_caso(change=None):
            linha = a3_falhas[a3_falhas['id_caso'] == a3_id_caso.value].iloc[0]
            a3_contexto.value = (
                f'<div style="padding:10px;border:1px solid #ccc">'
                f'<b>Pergunta:</b> {valor_texto(linha["pergunta"])}<br>'
                f'<b>Esperado:</b> {valor_texto(linha["comportamento_esperado"])}<br>'
                f'<b>Resposta observada:</b> {valor_texto(linha["resposta_observada"])}<br>'
                f'<b>Avaliação:</b> {valor_texto(linha["avaliacao"])}<br>'
                f'<b>Evidência:</b> {valor_texto(linha["evidencia"])}<br>'
                f'<b>Falha registrada:</b> {valor_texto(linha["descricao_falha"])}</div>'
            )
            a3_sintoma.value = valor_texto(linha['descricao_falha'])
            a3_evidencia.value = valor_texto(linha['evidencia'])

        def adicionar(_):
            with a3_saida_msg:
                clear_output()
                if not all([a3_sintoma.value.strip(), a3_evidencia.value.strip(), a3_h1.value.strip(), a3_h2.value.strip(), a3_justificativa.value.strip(), a3_verificacao.value.strip(), a3_resultado.value.strip()]):
                    print('❌ Preencha sintoma, evidência, duas hipóteses, justificativa, verificação e resultado esperado.')
                    return
                hipoteses = {'Hipótese 1': a3_h1.value.strip(), 'Hipótese 2': a3_h2.value.strip(), 'Hipótese 3': a3_h3.value.strip()}
                if not hipoteses[a3_escolha.value]:
                    print('❌ A hipótese escolhida está vazia.')
                    return
                linha = a3_falhas[a3_falhas['id_caso'] == a3_id_caso.value].iloc[0]
                identificador = f'D{len(a3_diagnosticos)+1:02d}'
                a3_diagnosticos.append({
                    'id_diagnostico': identificador,
                    'id_caso': a3_id_caso.value,
                    'pergunta': valor_texto(linha['pergunta']),
                    'comportamento_esperado': valor_texto(linha['comportamento_esperado']),
                    'resposta_observada': valor_texto(linha['resposta_observada']),
                    'avaliacao_original': valor_texto(linha['avaliacao']),
                    'sintoma_observado': a3_sintoma.value.strip(),
                    'evidencia_principal': a3_evidencia.value.strip(),
                    'hipotese_1': a3_h1.value.strip(),
                    'hipotese_2': a3_h2.value.strip(),
                    'hipotese_3': a3_h3.value.strip(),
                    'hipotese_escolhida': hipoteses[a3_escolha.value],
                    'justificativa_da_escolha': a3_justificativa.value.strip(),
                    'verificacao_planejada': a3_verificacao.value.strip(),
                    'resultado_esperado': a3_resultado.value.strip(),
                    'prioridade_diagnostico': a3_prioridade.value
                })
                print(f'✅ Diagnóstico {identificador} registrado.')
                a3_atualizar_tabela()

        def remover(_):
            with a3_saida_msg:
                clear_output()
                if not a3_diagnosticos:
                    print('Não há diagnósticos para remover.')
                    return
                removido = a3_diagnosticos.pop()
                print(f'🗑️ Diagnóstico {removido["id_diagnostico"]} removido.')
                a3_atualizar_tabela()

        def baixar(_):
            with a3_saida_msg:
                clear_output()
                if not a3_diagnosticos:
                    print('❌ Registre pelo menos um diagnóstico.')
                    return
                baixar_csv(pd.DataFrame(a3_diagnosticos), 'DIAGNOSTICO_NEUROQUIMICA.csv')
                print('✅ Ficha de diagnóstico criada.')

        a3_id_caso.observe(exibir_caso, names='value')
        a3_btn_add = widgets.Button(description='ADICIONAR DIAGNÓSTICO', button_style='success')
        a3_btn_remover = widgets.Button(description='REMOVER ÚLTIMO', button_style='warning')
        a3_btn_baixar = widgets.Button(description='BAIXAR DIAGNÓSTICO', button_style='info')
        a3_btn_add.on_click(adicionar)
        a3_btn_remover.on_click(remover)
        a3_btn_baixar.on_click(baixar)
        exibir_caso()
        display(widgets.VBox([
            a3_id_caso, a3_contexto, a3_sintoma, a3_evidencia, a3_h1, a3_h2,
            a3_h3, a3_escolha, a3_justificativa, a3_verificacao, a3_resultado,
            a3_prioridade, widgets.HBox([a3_btn_add, a3_btn_remover]),
            a3_saida_msg, widgets.HTML('<h4>Diagnósticos registrados</h4>'),
            a3_saida_tabela, a3_btn_baixar
        ]))
        a3_atualizar_tabela()

def a3_importar(_):
    global a3_falhas
    with a3_saida_upload:
        clear_output()
        enviados = files.upload()
        if not enviados:
            print('Nenhum arquivo selecionado.')
            return
        nome, conteudo = next(iter(enviados.items()))
        try:
            df = ler_csv_colab(conteudo)
        except Exception as erro:
            print(f'❌ Não foi possível abrir o arquivo: {erro}')
            return
        exigidas = ['id_caso', 'pergunta', 'comportamento_esperado', 'resposta_observada', 'avaliacao', 'evidencia', 'descricao_falha']
        faltantes = [c for c in exigidas if c not in df.columns]
        if faltantes:
            print('❌ O arquivo não parece ser o baseline da Aula 2.')
            print('Colunas ausentes: ' + ', '.join(faltantes))
            return
        a3_falhas = df[df['avaliacao'].isin(['Atende parcialmente', 'Não atende'])].copy()
        if a3_falhas.empty:
            print('⚠️ Nenhuma falha foi encontrada no arquivo.')
            return
        print(f'✅ {len(a3_falhas)} falha(s) carregada(s) de {nome}.')
        display(a3_falhas[['id_caso', 'pergunta', 'avaliacao', 'prioridade']])
        a3_criar_formulario()

a3_btn_importar = widgets.Button(description='CARREGAR BASELINE', button_style='primary')
a3_btn_importar.on_click(a3_importar)
display(widgets.VBox([a3_btn_importar, a3_saida_upload, a3_saida_form]))
```

```python
# AULA 5 — carregar resultados e construir o plano de evolução
if 'a5_melhorias' not in globals():
    a5_melhorias = []
a5_base = pd.DataFrame()
a5_comp = pd.DataFrame()

a5_saida_upload = widgets.Output()
a5_saida_form = widgets.Output()
a5_saida_msg = widgets.Output()
a5_saida_tabela = widgets.Output()
a5_saida_revisao = widgets.Output()

def a5_atualizar():
    with a5_saida_tabela:
        clear_output()
        if a5_melhorias:
            display(pd.DataFrame(a5_melhorias))
        else:
            print('Nenhuma melhoria foi adicionada.')
    with a5_saida_revisao:
        clear_output()
        if not a5_melhorias:
            return
        df = pd.DataFrame(a5_melhorias)
        print('REVISÃO DO PLANO')
        print(f'Total de melhorias: {len(df)}')
        print('\nPor prioridade:')
        print(df['prioridade'].value_counts().to_string())
        print('\nPor impacto:')
        print(df['impacto'].value_counts().to_string())
        sem_evidencia = (df['evidencia'].str.strip() == '').sum()
        sem_teste = (df['forma_de_teste'].str.strip() == '').sum()
        if sem_evidencia:
            print(f'\n⚠️ {sem_evidencia} melhoria(s) sem evidência.')
        if sem_teste:
            print(f'⚠️ {sem_teste} melhoria(s) sem forma de teste.')
        boas_primeiras = df[(df['impacto'] == 'Alto') & (df['esforco'].isin(['Baixo', 'Médio']))]
        if boas_primeiras.empty:
            print('\n💡 Procure identificar uma melhoria de alto impacto e esforço baixo ou médio.')

def a5_criar_formulario():
    with a5_saida_form:
        clear_output()
        ids = sorted(set(a5_base['id_caso']).union(set(a5_comp['id_caso'])))
        a5_id_caso = widgets.Dropdown(options=ids, description='Caso relacionado:', style={'description_width': '180px'})
        a5_contexto = widgets.HTML()
        a5_problema = widgets.Textarea(description='Problema:', style={'description_width': '180px'}, layout=widgets.Layout(width='950px', height='80px'))
        a5_evidencia = widgets.Textarea(description='Evidência:', style={'description_width': '180px'}, layout=widgets.Layout(width='950px', height='80px'))
        a5_melhoria = widgets.Textarea(description='Melhoria proposta:', style={'description_width': '180px'}, layout=widgets.Layout(width='950px', height='80px'))
        a5_impacto = widgets.Dropdown(options=['Alto', 'Médio', 'Baixo'], description='Impacto:', style={'description_width': '180px'})
        a5_esforco = widgets.Dropdown(options=['Baixo', 'Médio', 'Alto'], description='Esforço:', style={'description_width': '180px'})
        a5_risco = widgets.Dropdown(options=['Baixo', 'Médio', 'Alto'], description='Risco:', style={'description_width': '180px'})
        a5_prioridade = widgets.Dropdown(options=['Alta', 'Média', 'Baixa'], description='Prioridade:', style={'description_width': '180px'})
        a5_teste = widgets.Textarea(description='Como testar:', style={'description_width': '180px'}, layout=widgets.Layout(width='950px', height='80px'))
        a5_manter = widgets.Textarea(description='Critério para manter:', style={'description_width': '180px'}, layout=widgets.Layout(width='950px', height='70px'))
        a5_reverter = widgets.Textarea(description='Condição para reverter:', style={'description_width': '180px'}, layout=widgets.Layout(width='950px', height='70px'))
        a5_proximo = widgets.Textarea(description='Próximo passo:', style={'description_width': '180px'}, layout=widgets.Layout(width='950px', height='70px'))
        a5_confirmar_limpeza = widgets.Checkbox(value=False, description='Confirmo que desejo apagar todo o plano')

        def exibir(change=None):
            base = a5_base[a5_base['id_caso'] == a5_id_caso.value]
            comp = a5_comp[a5_comp['id_caso'] == a5_id_caso.value]
            partes = [f'<b>Caso:</b> {a5_id_caso.value}']
            if not base.empty:
                linha = base.iloc[0]
                partes.append(f'<b>Pergunta:</b> {valor_texto(linha.get("pergunta", ""))}')
                partes.append(f'<b>Falha inicial:</b> {valor_texto(linha.get("descricao_falha", ""))}')
            if not comp.empty:
                linha = comp.iloc[-1]
                partes.append(f'<b>Última decisão:</b> {valor_texto(linha.get("decisao", ""))}')
                partes.append(f'<b>Próximo passo registrado:</b> {valor_texto(linha.get("proximo_passo", ""))}')
            a5_contexto.value = '<div style="padding:10px;border:1px solid #ccc">' + '<br>'.join(partes) + '</div>'

        def adicionar(_):
            with a5_saida_msg:
                clear_output()
                obrigatorios = [a5_problema.value, a5_evidencia.value, a5_melhoria.value, a5_teste.value, a5_manter.value, a5_reverter.value, a5_proximo.value]
                if not all(v.strip() for v in obrigatorios):
                    print('❌ Preencha problema, evidência, melhoria, teste, critérios e próximo passo.')
                    return
                identificador = f'M{len(a5_melhorias)+1:02d}'
                a5_melhorias.append({
                    'id_melhoria': identificador,
                    'id_caso': a5_id_caso.value,
                    'problema_ou_oportunidade': a5_problema.value.strip(),
                    'evidencia': a5_evidencia.value.strip(),
                    'melhoria_proposta': a5_melhoria.value.strip(),
                    'impacto': a5_impacto.value,
                    'esforco': a5_esforco.value,
                    'risco': a5_risco.value,
                    'prioridade': a5_prioridade.value,
                    'forma_de_teste': a5_teste.value.strip(),
                    'criterio_para_manter': a5_manter.value.strip(),
                    'condicao_para_reverter': a5_reverter.value.strip(),
                    'proximo_passo': a5_proximo.value.strip()
                })
                print(f'✅ Melhoria {identificador} adicionada.')
                a5_atualizar()

        def remover(_):
            with a5_saida_msg:
                clear_output()
                if not a5_melhorias:
                    print('Não há melhorias para remover.')
                    return
                removida = a5_melhorias.pop()
                print(f'🗑️ Melhoria {removida["id_melhoria"]} removida.')
                a5_atualizar()

        def limpar(_):
            with a5_saida_msg:
                clear_output()
                if not a5_confirmar_limpeza.value:
                    print('⚠️ Marque a confirmação antes de apagar o plano.')
                    return
                a5_melhorias.clear()
                a5_confirmar_limpeza.value = False
                print('🗑️ Plano apagado.')
                a5_atualizar()

        def baixar(_):
            with a5_saida_msg:
                clear_output()
                if not a5_melhorias:
                    print('❌ Adicione pelo menos uma melhoria.')
                    return
                baixar_csv(pd.DataFrame(a5_melhorias), 'PLANO_EVOLUCAO_NEUROQUIMICA.csv')
                print('✅ Plano de evolução criado.')

        a5_id_caso.observe(exibir, names='value')
        a5_btn_add = widgets.Button(description='ADICIONAR MELHORIA', button_style='success')
        a5_btn_remover = widgets.Button(description='REMOVER ÚLTIMA', button_style='warning')
        a5_btn_limpar = widgets.Button(description='LIMPAR PLANO', button_style='danger')
        a5_btn_baixar = widgets.Button(description='BAIXAR PLANO', button_style='info')
        a5_btn_add.on_click(adicionar)
        a5_btn_remover.on_click(remover)
        a5_btn_limpar.on_click(limpar)
        a5_btn_baixar.on_click(baixar)
        exibir()
        display(widgets.VBox([
            a5_id_caso, a5_contexto, a5_problema, a5_evidencia, a5_melhoria,
            a5_impacto, a5_esforco, a5_risco, a5_prioridade, a5_teste,
            a5_manter, a5_reverter, a5_proximo,
            widgets.HBox([a5_btn_add, a5_btn_remover]),
            a5_confirmar_limpeza, a5_btn_limpar, a5_saida_msg,
            widgets.HTML('<h4>Plano de evolução</h4>'), a5_saida_tabela,
            widgets.HTML('<h4>Revisão</h4>'), a5_saida_revisao, a5_btn_baixar
        ]))
        a5_atualizar()

def a5_importar(_):
    global a5_base, a5_comp
    with a5_saida_upload:
        clear_output()
        enviados = files.upload()
        if len(enviados) < 2:
            print('❌ Selecione o baseline e a comparação ao mesmo tempo.')
            return
        base_encontrada = None
        comp_encontrada = None
        for nome, conteudo in enviados.items():
            try:
                df = ler_csv_colab(conteudo)
            except Exception:
                continue
            if {'id_caso', 'resposta_observada', 'avaliacao'}.issubset(df.columns):
                base_encontrada = df
            if {'id_comparacao', 'id_caso', 'decisao', 'nova_avaliacao'}.issubset(df.columns):
                comp_encontrada = df
        if base_encontrada is None or comp_encontrada is None:
            print('❌ Não foi possível identificar os dois arquivos.')
            return
        a5_base = base_encontrada
        a5_comp = comp_encontrada
        print(f'✅ Arquivos carregados: {len(a5_base)} caso(s) e {len(a5_comp)} comparação(ões).')
        display(a5_comp[['id_caso', 'avaliacao_anterior', 'nova_avaliacao', 'decisao', 'proximo_passo']])
        a5_criar_formulario()

a5_btn_importar = widgets.Button(description='CARREGAR OS DOIS ARQUIVOS', button_style='primary')
a5_btn_importar.on_click(a5_importar)
display(widgets.VBox([a5_btn_importar, a5_saida_upload, a5_saida_form]))
```
